# JAX Logistic Regression Demo

This notebook demonstrates how to train a basic logistic regression model using JAX, importing the logic from our `src/scientific_dev` package.

```python
import matplotlib.pyplot as plt
import jax.numpy as jnp
from scientific_dev.jax_model import generate_data, train_model

%matplotlib inline
```

## 1. Generate Data

```python
X, y = generate_data(n_samples=5000, n_features=20)
print(f"Data shape: X={X.shape}, y={y.shape}")
```

## 2. Train Model
We use the JIT-compiled update function from our package.

```python
params, history = train_model(X, y, epochs=2000, learning_rate=0.05)
```

## 3. Visualize Training

```python
plt.figure(figsize=(10, 6))
plt.plot(history)
plt.title('Training Loss Loop')
plt.xlabel('Steps (x100)')
plt.ylabel('Binary Cross Entropy')
plt.grid(True)
plt.show()
```
